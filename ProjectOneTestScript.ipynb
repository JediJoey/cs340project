```python
### 1. Import your CRUD Python module
from animal_shelter import AnimalShelter

### 2. Instantiate an instance of your class
# Use the aacuser account credentials for authentication.
username = "aacuser"
password = "SNHU1234"

shelter = AnimalShelter(username, password)
print("Connected to the aac database as '%s'." % username)
```

```python
### 3. Use your create function to create a new record in the aac database
new_animal = {
    "age_upon_outcome": "2 years",
    "animal_id": "A999999",
    "animal_type": "Dog",
    "breed": "Labrador Retriever Mix",
    "color": "Black",
    "name": "Test Dog",
    "outcome_type": "Adoption",
}

create_success = shelter.create(new_animal)
print("Create succeeded:", create_success)
```

```python
### 4. Use your read function to return records from the aac database
results = shelter.read({"animal_id": "A999999"})
print("Documents found:", len(results))
for document in results:
    print(document)
```

```python
### 5. Use your update function to update an existing record in the aac database
modified_count = shelter.update(
    {"animal_id": "A999999"},
    {"name": "Updated Test Dog", "outcome_type": "Transfer"},
)
print("Documents modified:", modified_count)

# Confirm the update by reading the record back.
for document in shelter.read({"animal_id": "A999999"}):
    print(document)
```

```python
### 6. Use your delete function to delete an existing record in the aac database
deleted_count = shelter.delete({"animal_id": "A999999"})
print("Documents deleted:", deleted_count)

# Confirm the delete: reading the record back should return an empty list.
print("Documents remaining:", len(shelter.read({"animal_id": "A999999"})))
```
